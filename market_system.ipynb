```python
import builtins

def update_stock(pid, qty_sold):
    with builtins.open(PRODUCTS_FILE, "r") as f:
        products = list(csv.DictReader(f))
    for p in products:
        if p["ID"] == pid:
            p["Stock"] = str(int(p["Stock"]) - qty_sold)
            break
    with builtins.open(PRODUCTS_FILE, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=["ID","Name","Category","Price","Stock"])
        writer.writeheader()
        writer.writerows(products)

def generate_bill():
    print("\n============================")
    print("        BILLING SYSTEM      ")
    print("============================")
    customer_name = input("Enter Customer Name: ")
    bill_no = f"BILL{datetime.now().strftime('%Y%m%d%H%M%S')}"
    date = datetime.now().strftime("%d-%m-%Y %H:%M")
    cart = []
    total = 0
    view_products()

    while True:
        pid = input("\nEnter Product ID (or done to finish): ")
        if pid.lower() == "done":
            break
        with builtins.open(PRODUCTS_FILE, "r") as f:
            products = list(csv.DictReader(f))
        found = False
        for p in products:
            if p["ID"] == pid:
                found = True
                stock = int(p["Stock"])
                if stock == 0:
                    print(f"'{p['Name']}' is OUT OF STOCK!")
                    break
                qty = int(input(f"Enter Quantity (Available: {stock}): "))
                if qty > stock:
                    print(f"Only {stock} available!")
                    break
                item_total = float(p["Price"]) * qty
                cart.append({
                    "bill_no": bill_no,
                    "date": date,
                    "pid": p["ID"],
                    "name": p["Name"],
                    "qty": qty,
                    "price": float(p["Price"]),
                    "total": item_total
                })
                total += item_total
                update_stock(pid, qty)
                print(f"Added {qty} x {p['Name']} = Rs.{item_total:.2f}")
                break
        if not found:
            print("Product ID not found!")

    if not cart:
        print("Cart is empty! No bill generated.")
        return

    gst = total * 0.05
    grand_total = total + gst

    print("\n" + "=" * 50)
    print("         GROCERY MARKET BILL             ")
    print("=" * 50)
    print(f"  Bill No  : {bill_no}")
    print(f"  Date     : {date}")
    print(f"  Customer : {customer_name}")
    print("=" * 50)
    print(f"{'Product':<20} {'Qty':>5} {'Price':>8} {'Total':>10}")
    print("-" * 50)
    for item in cart:
        print(f"{item['name']:<20} {item['qty']:>5} Rs.{item['price']:>7.2f} Rs.{item['total']:>9.2f}")
    print("-" * 50)
    print(f"{'SUBTOTAL':>36} Rs.{total:>9.2f}")
    print(f"{'GST (5%)':>36} Rs.{gst:>9.2f}")
    print("=" * 50)
    print(f"{'GRAND TOTAL':>36} Rs.{grand_total:>9.2f}")
    print("=" * 50)
    print("      Thank you for shopping!            ")
    print("=" * 50)

    with builtins.open(SALES_FILE, "a", newline="") as f:
        writer = csv.writer(f)
        for item in cart:
            writer.writerow([
                item["bill_no"], item["date"], item["pid"],
                item["name"], item["qty"], item["price"],
                item["total"]
            ])
    print(f"\nBill saved to {SALES_FILE}")

print("Billing Functions Loaded!")
```

```python
import builtins

# Add new customer
def add_customer():
    print("\n--- Add New Customer ---")
    cid    = f"CUST{datetime.now().strftime('%Y%m%d%H%M%S')}"
    name   = input("Enter Customer Name: ")
    phone  = input("Enter Phone Number: ")
    email  = input("Enter Email (press Enter to skip): ")
    joined = datetime.now().strftime("%d-%m-%Y")

    with builtins.open(CUSTOMERS_FILE, "a", newline="") as f:
        csv.writer(f).writerow([cid, name, phone, email, joined])
    print(f" Customer '{name}' added! ID: {cid}")

# View all customers
def view_customers():
    with builtins.open(CUSTOMERS_FILE, "r") as f:
        customers = list(csv.DictReader(f))

    if not customers:
        print(" No customers found!")
        return

    print(f"\n{'ID':<25} {'Name':<20} {'Phone':<15} {'Joined':<12}")
    print("-" * 75)
    for c in customers:
        print(f"{c['Customer ID']:<25} {c['Name']:<20} "
              f"{c['Phone']:<15} {c['Joined Date']:<12}")

# Search customer
def search_customer():
    print("\n--- Search Customer ---")
    keyword = input("Enter Name or Phone to search: ").lower()

    with builtins.open(CUSTOMERS_FILE, "r") as f:
        customers = list(csv.DictReader(f))

    results = [c for c in customers if
               keyword in c["Name"].lower() or
               keyword in c["Phone"]]

    if not results:
        print(" No customer found!")
        return

    print(f"\n{'ID':<25} {'Name':<20} {'Phone':<15} {'Joined':<12}")
    print("-" * 75)
    for c in results:
        print(f"{c['Customer ID']:<25} {c['Name']:<20} "
              f"{c['Phone']:<15} {c['Joined Date']:<12}")

# Delete customer
def delete_customer():
    view_customers()
    cid = input("\nEnter Customer ID to delete: ")

    with builtins.open(CUSTOMERS_FILE, "r") as f:
        customers = list(csv.DictReader(f))

    new_customers = [c for c in customers if c["Customer ID"] != cid]

    if len(new_customers) == len(customers):
        print(" Customer ID not found!")
        return

    with builtins.open(CUSTOMERS_FILE, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=["Customer ID", "Name",
                                                "Phone", "Email", "Joined Date"])
        writer.writeheader()
        writer.writerows(new_customers)
    print(" Customer deleted successfully!")

# View purchase history
def view_purchase_history():
    print("\n--- Customer Purchase History ---")
    keyword = input("Enter Customer Name to search: ").lower()

    with builtins.open(CUSTOMERS_FILE, "r") as f:
        customers = list(csv.DictReader(f))

    matched = [c for c in customers if keyword in c["Name"].lower()]

    if not matched:
        print(" Customer not found!")
        return

    customer = matched[0]
    print(f"\n Customer : {customer['Name']}")
    print(f"   Phone    : {customer['Phone']}")
    print(f"   Joined   : {customer['Joined Date']}")

    with builtins.open(SALES_FILE, "r") as f:
        sales = list(csv.DictReader(f))

    customer_sales = [s for s in sales if
                      customer["Name"].lower() in s["Bill No"].lower()]

    if not customer_sales:
        print(" No purchase history found!")
        return

    print(f"\n{'Bill No':<25} {'Date':<20} {'Product':<20} {'Qty':>5} {'Total':>10}")
    print("-" * 85)

    grand_total = 0
    for s in customer_sales:
        print(f"{s['Bill No']:<25} {s['Date']:<20} "
              f"{s['Product Name']:<20} {s['Qty']:>5} "
              f"Rs.{float(s['Total']):>9.2f}")
        grand_total += float(s["Total"])

    print("-" * 85)
    print(f"{'Total Spent':>72} Rs.{grand_total:>9.2f}")

print(" Customer Functions Loaded!")
```

```python
def customer_menu():
    while True:
        print("\n============================")
        print("       CUSTOMER MENU      ")
        print("============================")
        print("1. Add Customer")
        print("2. View All Customers")
        print("3. Search Customer")
        print("4. Delete Customer")
        print("5. View Purchase History")
        print("6. Back to Main Menu")

        choice = input("Choose option: ")

        if choice == "1":
            add_customer()
        elif choice == "2":
            view_customers()
        elif choice == "3":
            search_customer()
        elif choice == "4":
            delete_customer()
        elif choice == "5":
            view_purchase_history()
        elif choice == "6":
            print(" Going back to Main Menu...")
            break
        else:
            print(" Invalid option! Choose 1-6")

# Test it!
customer_menu()
```

```python
import builtins

# Daily sales report
def daily_sales_report():
    with builtins.open(SALES_FILE, "r") as f:
        sales = list(csv.DictReader(f))

    if not sales:
        print(" No sales data found!")
        return

    today = datetime.now().strftime("%d-%m-%Y")
    today_sales = [s for s in sales if today in s["Date"]]

    print("\n" + "=" * 55)
    print("           DAILY SALES REPORT               ")
    print(f"             Date: {today}                   ")
    print("=" * 55)

    if not today_sales:
        print(" No sales recorded today!")
        return

    print(f"{'Bill No':<25} {'Product':<15} {'Qty':>5} {'Total':>10}")
    print("-" * 55)

    total_revenue = 0
    total_items   = 0

    for s in today_sales:
        print(f"{s['Bill No']:<25} {s['Product Name']:<15} "
              f"{s['Qty']:>5} Rs.{float(s['Total']):>9.2f}")
        total_revenue += float(s["Total"])
        total_items   += int(s["Qty"])

    print("=" * 55)
    print(f"  Total Items Sold   : {total_items}")
    print(f"  Total Revenue      : Rs.{total_revenue:.2f}")
    print(f"  GST Collected (5%) : Rs.{total_revenue * 0.05:.2f}")
    print(f"  Grand Total        : Rs.{total_revenue * 1.05:.2f}")
    print("=" * 55)

# Best selling products
def best_selling_products():
    with builtins.open(SALES_FILE, "r") as f:
        sales = list(csv.DictReader(f))

    if not sales:
        print(" No sales data found!")
        return

    product_sales = {}
    for s in sales:
        name    = s["Product Name"]
        qty     = int(s["Qty"])
        revenue = float(s["Total"])

        if name in product_sales:
            product_sales[name]["qty"]     += qty
            product_sales[name]["revenue"] += revenue
        else:
            product_sales[name] = {"qty": qty, "revenue": revenue}

    sorted_products = sorted(product_sales.items(),
                             key=lambda x: x[1]["qty"],
                             reverse=True)

    print("\n" + "=" * 55)
    print("         BEST SELLING PRODUCTS              ")
    print("=" * 55)
    print(f"{'Rank':<6} {'Product':<20} {'Qty Sold':>10} {'Revenue':>12}")
    print("-" * 55)

    for rank, (name, data) in enumerate(sorted_products, 1):
        print(f"{rank:<6} {name:<20} {data['qty']:>10} "
              f"Rs.{data['revenue']:>11.2f}")

    print("=" * 55)
    print(f"   Best Seller : {sorted_products[0][0]}")
    print("=" * 55)

# Low stock report
def low_stock_report():
    with builtins.open(PRODUCTS_FILE, "r") as f:
        products = list(csv.DictReader(f))

    low = [p for p in products if int(p["Stock"]) < 5]
    out = [p for p in products if int(p["Stock"]) == 0]

    print("\n" + "=" * 50)
    print("           STOCK ALERT REPORT              ")
    print("=" * 50)

    if out:
        print("\n OUT OF STOCK:")
        for p in out:
            print(f"    {p['Name']} (ID: {p['ID']})")

    if low:
        print("\n LOW STOCK (Less than 5):")
        print(f"{'ID':<6} {'Name':<20} {'Stock':>8}")
        print("-" * 36)
        for p in low:
            print(f"{p['ID']:<6} {p['Name']:<20} {p['Stock']:>8} ")
    else:
        print(" All products have sufficient stock!")

    print("=" * 50)

# Overall summary report
def summary_report():
    with builtins.open(SALES_FILE, "r") as f:
        sales = list(csv.DictReader(f))

    with builtins.open(PRODUCTS_FILE, "r") as f:
        products = list(csv.DictReader(f))

    with builtins.open(CUSTOMERS_FILE, "r") as f:
        customers = list(csv.DictReader(f))

    total_revenue = sum(float(s["Total"]) for s in sales)
    total_items   = sum(int(s["Qty"]) for s in sales)
    total_bills   = len(set(s["Bill No"] for s in sales))
    low_stock     = len([p for p in products if int(p["Stock"]) < 5])

    print("\n" + "=" * 50)
    print("          OVERALL SUMMARY REPORT            ")
    print("=" * 50)
    print(f"   Total Customers    : {len(customers)}")
    print(f"   Total Products     : {len(products)}")
    print(f"    Low Stock Items    : {low_stock}")
    print(f"   Total Bills        : {total_bills}")
    print(f"   Total Items Sold   : {total_items}")
    print(f"   Total Revenue      : Rs.{total_revenue:.2f}")
    print(f"   GST (5%)           : Rs.{total_revenue * 0.05:.2f}")
    print(f"   Grand Total        : Rs.{total_revenue * 1.05:.2f}")
    print("=" * 50)

print("Reports Functions Loaded!")
```
